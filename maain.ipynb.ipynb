```python
from datetime import datetime
import random
```

```python
class Notification:
    def __init__(self, message):
        # Validate message
        if not isinstance(message, str) or not message.strip():
            raise ValueError("Message cannot be empty.")

        self.message = message
        self.created_at = datetime.now()

    def send(self):
        # Base send method
        print(f"Notification sent: {self.message}")

    def __str__(self):
        # Return readable string
        return f"[{self.created_at.strftime('%Y-%m-%d %H:%M:%S')}] {self.message}"

    def __len__(self):
        # Return message length
        return len(self.message)


class EmailNotification(Notification):
    def send(self):
        # Send email
        print(f"Email sent: {self.message}")


class SMSNotification(Notification):
    def send(self):
        # Send SMS
        print(f"SMS sent: {self.message}")


class PushNotification(Notification):
    def send(self):
        # Send push notification
        print(f"Push notification sent: {self.message}")
```

```python
def generate_message():
    # Return random message
    messages = [
        "Your order has been shipped.",
        "Your password was changed successfully.",
        "You have a new friend request.",
        "Your subscription will expire soon.",
        "A new login to your account was detected."
    ]
    return random.choice(messages)
```

```python
email_notification = EmailNotification(generate_message())
sms_notification = SMSNotification(generate_message())
push_notification = PushNotification(generate_message())

notifications = [email_notification, sms_notification, push_notification]

for notification in notifications:
    print(notification)
```

```python
for notification in notifications:
    notification.send()
```

```python
print(email_notification)
print("Length using len(object):", len(email_notification))
print("Length using len(message):", len(email_notification.message))
```
