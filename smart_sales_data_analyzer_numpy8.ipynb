# Project: Smart Sales Data Analyzer Numpy

### Project Objective
The goal of this project is to analyze a 7-day sales dataset for a retail store using **NumPy**. By representing daily sales across multiple products in a matrix format, we aim to perform efficient data manipulation and statistical analysis.

### Why NumPy?
1. **Performance:** NumPy is significantly faster than standard Python lists because it stores data in contiguous memory blocks.
2. **Vectorization:** It allows us to perform mathematical operations on entire arrays without using explicit `for` loops, making the code cleaner and faster.
3. **Optimized Functions:** It provides built-in statistical methods (sum, mean, max, min) that are highly optimized for large datasets.

### Core Concepts Covered
* **Array Creation:** Using `np.random.randint` to simulate datasets.
* **Structural Inspection:** Understanding `shape`, `ndim`, and `dtype`.
* **Indexing & Slicing:** Accessing specific days or products within the dataset.
* **Statistical Methods:** Computing total, average, and extreme values.
* **Axis Operations:** Mastering the difference between `axis=0` (column-wise) and `axis=1` (row-wise) calculations.

```python
import numpy as np

# Creating a 7x5 matrix with random sales data between 10 and 500
# 7 rows represent days, 5 columns represent products
sales_data = np.random.randint(10, 500, size=(7, 5))

# Displaying the generated matrix
print(" 7-Day Sales Matrix ")
print(sales_data)
```

```python
# Printing the key properties of our sales matrix
print(f"Matrix Shape (Days, Products): {sales_data.shape}")
print(f"Number of Dimensions: {sales_data.ndim}")
print(f"Data Type of Elements: {sales_data.dtype}")
print(f"Total Number of Entries: {sales_data.size}")

# Let's see the matrix again to be sure
print("7-Day Sales Matrix ")
print(sales_data)
```

```python
# 1. Sales for the 3rd day (Index 2)
third_day_sales = sales_data[2, :]
print(f"Sales for the 3rd day: {third_day_sales}")

# 2. Weekly sales for the last product (Last column)
last_product_sales = sales_data[:, -1]
print(f"Weekly sales for the last product: {last_product_sales}")

# 3. Slicing: Sales for the first 3 days
first_three_days = sales_data[0:3, :]
print(" Sales for the first 3 days")
print(first_three_days)
```

```python

total_sales = sales_data.sum()
average_sales = sales_data.mean()
max_value = sales_data.max()
min_value = sales_data.min()

print(f"Total Weekly Sales: {total_sales}")
print(f"Daily Average Sales (Overall): {average_sales:.2f}") 
print(f"Highest Sale Recorded: {max_value}")
print(f"Lowest Sale Recorded: {min_value}")
```

```python
# Calculating total sales per day (Summing across columns for each row)
# axis=1: Performs the operation horizontally (row-wise)
daily_totals = sales_data.sum(axis=1)

# Calculating total sales per product (Summing across rows for each column)
# axis=0: Performs the operation vertically (column-wise)
product_totals = sales_data.sum(axis=0)

print(" Total Sales Per Day ")
print(daily_totals)

print("Total Sales Per Product")
print(product_totals)

# EXPLANATION:
# axis=1 collapses the columns to give us a total for each row (Day).
# axis=0 collapses the rows to give us a total for each column (Product).
```

```python
# 1. Reshaping the 7x5 matrix into 5x7
# (Note: The total number of elements must remain the same: 35)
reshaped_sales = sales_data.reshape(5, 7)

# 2. Transposing the matrix (Swapping rows and columns)
transposed_sales = sales_data.T # or sales_data.transpose()

# 3. Flattening the matrix (Converting 2D table into a 1D array)
flattened_sales = sales_data.flatten()

print(f"Original Shape: {sales_data.shape}")
print(f"Reshaped Matrix (5x7):{reshaped_sales}")
print(f"\nTransposed Matrix (5x7):{transposed_sales}")
print(f"\nFlattened Array (Total 35 elements):{flattened_sales}")
```

```python
# 1. Total Weekly Sales
total_weekly = sales_data.sum()

# 2. Finding the best performing product
# np.argmax returns the index of the maximum value
best_product_index = np.argmax(product_totals) 
# We add +1 because indices start at 0, but we want to say "Product 1, Product 2..."
best_product_name = f"Product {best_product_index + 1}"

# 3. Finding the best day of the week
best_day_index = np.argmax(daily_totals)
days_of_week = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
best_day_name = days_of_week[best_day_index]

print("--- FINAL SALES SUMMARY ---")
print(f"Total Weekly Sales: {total_weekly} units")
print(f"Best Performing Product: {best_product_name} (Total: {product_totals[best_product_index]})")
print(f"Best Day of the Week: {best_day_name} (Total: {daily_totals[best_day_index]})")
```

1. Why NumPy Over Python Lists?
In this project, we utilized NumPy arrays instead of standard Python lists for several critical reasons:

Memory Efficiency: NumPy arrays are stored in contiguous memory blocks, making them much more compact than Python lists.

Speed: Operations on NumPy arrays are implemented in C, allowing for significantly faster computations, especially as the dataset grows.

Vectorization: We performed calculations (like adding taxes or finding totals) across the entire dataset without writing a single for loop. This "vectorized" approach reduces code complexity and potential human error.

2. The Importance of "Axis" in Data Analysis
The concept of an Axis is fundamental when working with multi-dimensional data (matrices):

Axis 1 (Row-wise): By calculating across Axis 1, we collapsed the columns to find the total sales for each Day. This is essential for understanding temporal trends.

Axis 0 (Column-wise): By calculating across Axis 0, we collapsed the rows to find the total sales for each Product. This helps identify which items are performing best in the market.

3. Structural Flexibility: Reshape & Transpose
Data often comes in a format that isn't ideal for every type of analysis.

Reshaping allowed us to change the structure of our 35 data points (e.g., from 7x5 to 5x7) without losing or duplicating information.

Transposing was crucial for flipping our perspective, turning products into rows and days into columns, which is often required for specific matrix multiplications or advanced statistical modeling.

4. Conclusion
NumPy transformed a raw collection of random numbers into a structured, actionable sales report. By using built-in methods like sum(), mean(), and argmax(), we successfully identified the most profitable day and the top-selling product with minimal code, proving why NumPy is the backbone of the Python Data Science ecosystem.
